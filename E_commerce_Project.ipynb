```python
import pandas as pd
import plotly.express as px #visualization
import plotly.graph_objects as go  #advanced graphs
import plotly.io as pio #for graph templates
import plotly.colors as colors
pio.templates.default="plotly_dark"
```

```python
data=pd.read_csv("Sample - Superstore.csv",encoding='latin-1') #latin reads the special characters
```

```python
data.head()
```

```python
data.describe() #returns the descriptive statistics of integers
```

```python
data.info() #gives more information of columns,on null values and its datatypes
```

# **To convert dtype to data&time dtype**



```python
data['Order Date']=pd.to_datetime(data['Order Date'])
data['Ship Date']=pd.to_datetime(data['Ship Date'])
```

```python
data.info()
```

```python
data.head(10) # first 10 indeces -- by default its only 5
```

# **Creating New columns**

```python
data['Order Month'] = data['Order Date'].dt.month
data['Order Year'] = data['Order Date'].dt.year
data['Order Day of Week'] = data['Order Date'].dt.dayofweek
```

```python
data.tail(10) #gives last 10 indeces -- be defalut it gives 5 only
```

```python
data[10:20] # to access inbetween data we use slicing
```

## **Monthly Sales Analysis**

```python
data
```

# **Sales By Category**

```python
sales_by_month=data.groupby('Order Month')['Sales'].sum().reset_index()
```

```python
sales_by_month
```

```python
fig=px.line(sales_by_month,
            x='Order Month',
            y='Sales',
            title='Monthly Sales Analysis')
fig.show()
```

```python
sales_by_category=data.groupby('Category')['Sales'].sum().reset_index()
```

```python
sales_by_category
```

```python
fig2=px.bar(sales_by_category,x='Category',
            y='Sales',title='Category Wise Sales')
fig2.show()
```

```python
fig2=px.pie(sales_by_category,values='Sales',names='Category',title='Category Wise Sales')
fig2.show()
```

```python
fig2=px.pie(sales_by_category,
            values='Sales',
            names='Category',
            hole=0.5,
            color_discrete_sequence=px.colors.qualitative.Prism)

fig2.update_traces(textposition='inside',textinfo='percent+label')
fig2.update_layout(title_text='Sales Analysis by Category',title_font=dict(size=24))
fig2.show()
```

# **Sales Analysis by SubCategories**

```python
sales_by_subcategory=data.groupby(['Category','Sub-Category'])['Sales'].sum().reset_index()
sales_by_subcategory
```

```python
fig3=px.bar(sales_by_subcategory,
            x='Sub-Category',
            y='Sales',
            title='Sales Analysis by SubCategory',
            color='Category',
            barmode='group')
fig3.show()
```

# **Category wise Monthly Profit**

```python
monthly_profit_by_category=data.groupby(['Order Month','Category'])['Profit'].sum().reset_index()
monthly_profit_by_category
```

```python
fig4=px.bar(monthly_profit_by_category,
            x='Order Month',
            y='Profit',
            title='Monthly Sales by Profit',
            color='Category',
            barmode='group'
            )
fig4.update_layout(
    xaxis=dict(tickmode='linear'))  # to get all the numbers on x-axis
fig4.show()
```

# **Profit & Sales Analysis by Customer Segments**

```python
profit_sales_by_segments=data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
profit_sales_by_segments
```

```python
fig8=go.Figure()
fig8.add_trace(go.Bar(x=profit_sales_by_segments['Segment'],
                      y=profit_sales_by_segments['Sales'],
                      name='Sales',
                      marker_color='blue'))
fig8.add_trace(go.Bar(x=profit_sales_by_segments['Segment'],
                      y=profit_sales_by_segments['Profit'],
                      name='Profit',
                      marker_color='green'))
fig8.update_layout(title_text='Profit & Sales Analysis by Customer Segments',
                  xaxis_title='Customer Segments',
                  yaxis_title='Amount')
fig8.show()
```

# **Sales to Profit Ratio**

```python
sales_profit_ratio_by_segment=data.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_profit_ratio_by_segment['Sales_to_Profit_Ratio']=sales_profit_ratio_by_segment['Sales']/sales_profit_ratio_by_segment['Profit']
print(sales_profit_ratio_by_segment)
```

```python
fig10=px.pie(sales_profit_ratio_by_segment,
             values='Sales_to_Profit_Ratio',
             names='Segment',title='Sales to Profit Ratio by Segment')
fig10.update_traces(textposition='inside',textinfo='percent+label')
fig10.show()
```
